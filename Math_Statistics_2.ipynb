```python
def get_max_likelihood(destribution):
  sum=0
  n=len(destribution)
  for i in range(n):
    sum+=destribution[i]*destribution[i]
  res=np.sqrt((1/(2.0*n))*sum)
  return res
```

```python
def flatten(data):
  if isinstance(data[0], list) or isinstance(data[0], np.ndarray):
    return [x for sublist in data for x in sublist]
  else:
    return data
```

```python
def get_function_of_likelihood_log(data, sigma_vals):
  data=flatten(data)
  data=np.array(data)
  n=len(data)
  print(n)
  sum_ln=np.sum(np.log(data))
  sum_x2=np.sum(data*data)

  print('sum_ln=',sum_ln)
  print('sum_x2=',sum_x2)
  resArr = []
  for sigma in sigma_vals:
    if sigma<=0:
      resArr.append(-np.inf)
    else:
      res=(-2*n*np.log(sigma))+sum_ln-(1/(2*sigma*sigma))*sum_x2
      resArr.append(res)
  sigma_mmp=np.sqrt(sum_x2/(2*n))
  return resArr, sigma_mmp
```

```python
def get_function_of_likelihood_log_v2(data, sigma_vals):
  n=len(data) * len(data[0])
  print(n)
  #sum_ln=np.sum(np.log(data))
  #sum_x2=np.sum(data*data)
  sum_ln = 0
  sum_x2 = 0
  for r in data:
    for x in r:
      sum_ln += np.log(x)
      sum_x2 += x*x

  print('sum_ln=',sum_ln)
  print('sum_x2=',sum_x2)

  resArr = []
  for sigma in sigma_vals:
    if sigma<=0:
      resArr.append(-np.inf)
    else:
      res=(-2*n*np.log(sigma))+sum_ln-(1/(2*sigma*sigma))*sum_x2
      resArr.append(res)

  sigma_mmp=np.sqrt(sum_x2/(2*n))
  return resArr, sigma_mmp
```
