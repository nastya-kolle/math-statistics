моделируем 1000 выборок нормального распределения объема n=10
сразу вычисляем выборочное среднее и выборочную дисперсию для каждой из выборок
все храним в списке

```python
s=[]
m=[]
d=[]
n=10
for i in range(1000):
  a=np.random.rayleigh(scale=2,size=n)
  s.append(a)
  m.append(np.mean(a))#вычисление выборочного среднего
  d.append(np.var(a, ddof=1))#вычисление исправленной выборочной дисперсии
```

```python
import matplotlib.pyplot as plt
f1=np.sqrt(2*np.pi)
plt.plot(range(0, len(m)), m,color='b', linewidth=1)
plt.axhline(y=np.sqrt(2*np.pi),color='r')
plt.show()
```

```python
plt.plot(range(0, len(d)),d,color='r', linewidth=1 )
plt.axhline(y=8-2*np.pi,color='b')
plt.show()
```
